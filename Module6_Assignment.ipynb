```python
#Task 1
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv"
df = pd.read_csv(url)
df.head()
df.info()
df.describe()
df.isnull().sum()
```

```python
#Task 2
#Cabin
cabin_missing_percentaeg = df["Cabin"].isnull().mean() * 100
print(f"Cabin missing percentage: {cabin_missing_percentage:.2f}%")
"""Around 77% of the Cabin values are missing, which is extremely high.
For this keeping the column would provide limited useful information,so we can drop it."""

#Embark
embarked_mode = df["Embarked"].mode()[0]
print("Most frequent port:",embarked_mode)
df["Embarked"]=df["Embarked"].fillna(embarked_mode)

#Age
age_median = df["Age"].median()
print("Median age:", age_median)
df["Age"] = df["Age"].fillna(age_median)
```

```python
#Task 3
import seaborn as sns
import matplotlib.pyplot as plt

survival_rate = df["Survived"].mean() * 100
print(f"Overall survival rate: {survival_rate:.2f}%")

sns.countplot(x="Survived", data=df)

plt.title("Survival Distribution")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")

plt.show()

sns.countplot(x="Pclass", data=df)

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()

plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=30)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
#Task 4

# SURVIVAL BY SEX

sex_survival = pd.crosstab(df["Sex"], df["Survived"])

print("Survival Count by Sex:")
print(sex_survival)

sex_survival_rate = df.groupby("Sex")["Survived"].mean() * 100

print("\nSurvival Rate by Sex:")
print(sex_survival_rate.round(2))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")

plt.show()

"Female passengers had a significantly higher survival rate than male passengers."

# SURVIVAL BY CLASS (Pclass)

class_survival_rate = df.groupby("Pclass")["Survived"].mean() * 100
print("Survival Rate by Pclass:")
print(class_survival_rate.round(2))


sns.barplot(
    x=class_survival_rate.index,
    y=class_survival_rate.values
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.show()
"Yes,there is a clear relationship between ticket class and survival probability.1st class had the highest survival rate, while 3rd class had the lowest survival rate."

# SURVIVAL BY AGE
sns.histplot(
    data=df,
    x="Age",
    hue="Survived",
    bins=30,
    kde=True
)

plt.title("Age Distribution of Survivors vs Non-Survivors")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()

"The children generally had relatively better survival chances than many adults,while elderly passengers tended to have lower survival chances.However, age alone was not as strong a predictor of survival as Sex and Pclass"
```
