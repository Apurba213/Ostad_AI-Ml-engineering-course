```python
#Question 1: Loop Basics (for loop)
for i in range(10):
  print(f"i={i+1}")
```

```python
#Question 2: Loop with Condition

#Using a for loop, print only the even numbers between 1 and 20
for i in range(1,21):
  if i%2==0:
    print(i)
```

```python
#Question 3: String Manipulation

string1=input("Enter a string\n")
print(f"\nThe Upper case={string1.upper()}\n")
print(f"Len={len(string1)}\n")
print(f"Reverse={string1[::-1]}")

```

```python
#Question 4: String Processing with Loop
string1=input("Enter a string\n")
cnt=0
for i in string1:
  if i == 'a' or i == 'e' or i == 'i' or i == 'o' or i == 'u':
    cnt+=1
print(cnt)
```

```python
#Question 5: List Comprehension
numbers = [10, 25, 30, 45, 50, 65, 70]
number1=[]
#Create a new list containing only numbers greater than 40
for i in numbers:
  if i>40:
    number1.append(i)

print("New list=",number1)
```
