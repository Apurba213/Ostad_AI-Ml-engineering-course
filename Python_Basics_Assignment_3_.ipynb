```python
#Question 1: Enumerate with List
fruits = ["apple", "banana", "cherry", "mango"]
for index,fruit in enumerate(fruits):
    print(index,fruit)
```

```python
#Question 2: Zip with Two Lists
students = ["Rahim", "Karim", "Ayesha"]
scores = [85, 90, 78]
for student,score in zip(students,scores):
    print(f"{student} scored {score}")
```

```python
#Question 3: Tuple and Loop

new_tuple = ("Math", "Science", "English", "History", "Art")
for subject in new_tuple:
    print(subject)

```

```python
#Question 4: Set Operations
#Given the sets:

set1 = {1, 2, 3, 4, 5}

set2 = {4, 5, 6, 7}
"""
Print the common elements using set operation
Print the unique elements of set1
"""
common_elements = set1.intersection(set2)
print(f"Common elements are {common_elements}")
unique_elements = set1.union(set2)
print(f"Unique elements are {unique_elements}")

```

```python
#Given the dictionary:

marks = {

    "Math": 80,

    "Physics": 75,

    "Chemistry": 85

}
for index, (subject, mark) in enumerate(marks.items(), start=1):
    print(f"{index} {subject}: {mark}")
```
