```python
#Question 1:
#A:
#Ask the user for input
num1 = int(input("Enter a number: "))
num2 = int(input("Enter another number: "))
#B:
#Addition
print("\nAddition=",num1 + num2)
#Subtraction
print("\nSubtraction=",num1 - num2)
#Multiplication
print("\nMultiplication=",num1 * num2)
#Division
print("\nDivision=",num1 / num2)
```

```python
# Question 2
#A
#Ask the user for their information
name = input("Enter your name: ")
city = input("Enter your city: ")
hobby = input("Enter your favourite hobby: ")
#B
print(f"Hello {name}! You live in {city} and you enjoy {hobby}.")

```

```python
#Question 3
#A
#Ask the user to enter the temperature in Celsius and convert to float
celsius = float(input("Enter temperature in Celsius: "))
#B
#Convert Celsius to Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32
#Display the result
print(f"Temperature in Fahrenheit is: {fahrenheit}°F")
```

```python
#Question 4
#Ask the user for item details
#A
item = input("Enter the item name: ")
price = float(input("Enter the price: "))
quantity = int(input("Enter the quantity: "))
#B
#Calculate the total bill
total = price * quantity

print(f"Total Bill for {item} = {total}")
```
