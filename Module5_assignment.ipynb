```python
class ShoppingCart:
    # Part A: Constructor
    def __init__(self, customer_name, items=None):
        self.customer_name = customer_name
        if items is None:
            self.items = []
        else:
            self.items = items
    # Part B: Add item
    def add_item(self, price):
        if price > 0:
            self.items.append(price)
        else:
            print("Price must be a positive number.")
    # Part B: Remove item
    def remove_item(self, price):
        if price in self.items:
            self.items.remove(price)
        else:
            print("Item does not exist in the cart.")
    # Part B: Calculate total
    def calculate_total(self):
        total = 0
        for price in self.items:
            total += price
        return total
    # Part B: Apply discount
    def apply_discount(self):
        total = self.calculate_total()
        if total >= 3000:
            final_amount = total * 0.9
        else:
            final_amount = total
        return final_amount
    # Part C: Display cart
    def display_cart(self):
        print("\n----- Shopping Cart Details -----")
        print("Customer Name:", self.customer_name)
        print("Items in the Cart:", self.items)
        print("Total Amount:", self.calculate_total())
        print("Final Amount after Discount:", self.apply_discount())
        print("Number of Items in the Cart:", len(self.items))
        print("---------------------------------")
    # Part C: Clear cart
    def clear_cart(self):
        self.items = []
        print("Cart has been cleared.")
    # Part D: Operator Overloading
    def __add__(self, other):
        new_customer_name = (
            self.customer_name + " & " + other.customer_name)
        new_items = self.items + other.items
        return ShoppingCart(new_customer_name, new_items)
# Part E: Object Usage & Output
# 1. Create two ShoppingCart objects
cart1 = ShoppingCart("Apurba")
cart2 = ShoppingCart("Rahim")
# 2. Add items to the first cart
cart1.add_item(1000)
cart1.add_item(1500)
cart1.add_item(2000)
# Remove one item from the first cart
cart1.remove_item(1500)
# Add items to the second cart
cart2.add_item(500)
cart2.add_item(1000)
cart2.add_item(2500)
# Remove one item from the second cart
cart2.remove_item(500)
# 3. Display individual cart details
print("CART 1")
cart1.display_cart()
print("\nCART 2")
cart2.display_cart()
# 4. Combine both carts using +
combined_cart = cart1 + cart2
# 5. Display combined cart details
print("\nCOMBINED CART")
combined_cart.display_cart()
```
